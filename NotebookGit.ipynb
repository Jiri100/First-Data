### Data Type Conversion


```python
nakazeni = df['kumulativni_pocet_nakazenych']
```

```python
datum = pd.to_datetime(df['datum'])
```

```python
plt.plot(datum, nakazeni, linestyle = '-')
plt.xlabel('Datum')
plt.ylabel('Počet nakažených')
plt.title('Počet nakažených v čase')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```
